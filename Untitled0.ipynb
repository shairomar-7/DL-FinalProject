```python
import numpy as np, librosa, shutil
from tqdm import tqdm
src, dst = '/content/PianoVAM', f'{DRIVE}/data/PianoVAM'
for sub in ['Landmarks', 'Audio16k']:
    os.makedirs(f'{dst}/{sub}', exist_ok=True)

ids = sorted(v['record_time'] for v in meta.values())
has_a = {i for i in ids if os.path.exists(f'{src}/Audio/{i}.wav')}
has_h = {i for i in ids if os.path.exists(f'{src}/Handskeleton/{i}.json')}
print('missing audio:', sorted(set(ids) - has_a))
print('missing hands:', sorted(set(ids) - has_h))

# Hand landmarks -> float16 array (T frames, 2 hands [L, R], 21 joints, xyz); NaN = hand not detected
for i in tqdm(sorted(has_h), desc='landmarks'):
    out = f'{dst}/Landmarks/{i}.npz'
    if os.path.exists(out): continue
    d = json.load(open(f'{src}/Handskeleton/{i}.json'))
    T = max(int(k) for k in d) + 1
    X = np.full((T, 2, 21, 3), np.nan, dtype=np.float16)
    for t in range(T):
        f = d.get(str(t)) or {}
        for h, name in enumerate(['Left', 'Right']):
            if f.get(name) is not None:
                X[t, h] = f[name]
    np.savez_compressed(out, landmarks=X)

# Audio -> 16 kHz mono
for i in tqdm(sorted(has_a), desc='audio'):
    out = f'{dst}/Audio16k/{i}.wav'
    if os.path.exists(out): continue
    y, _ = librosa.load(f'{src}/Audio/{i}.wav', sr=16000, mono=True)
    sf.write(out, y, 16000, subtype='PCM_16')

for f in ['MIDI', 'TSV']:
    shutil.copytree(f'{src}/{f}', f'{dst}/{f}', dirs_exist_ok=True)
for f in ['metadata.json', 'README.md']:
    shutil.copy(f'{src}/{f}', dst)

!du -sh "{dst}"/*
```
